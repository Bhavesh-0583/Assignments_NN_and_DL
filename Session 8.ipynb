### 1. Numerical Gradient Using Central Difference Method

```python
def compute_gradient(f, x):
    h = 1e-5   
    gradient = (f(x + h) - f(x - h)) / (2 * h)   
    return gradient
def f(x):
    return x ** 2
x = 2
print("Gradient at x =", x, "is:", compute_gradient(f, x))
```

### 2. Partial Derivatives of f(x,y) = x² + y²

```python
def partial_derivatives(x, y):
    dx = 2 * x
    dy = 2 * y    
    return dx, dy
x = 3
y = 4
dx, dy = partial_derivatives(x, y)
print("Partial derivative with respect to x:", dx)
print("Partial derivative with respect to y:", dy)
```

### 3. One Step of Gradient Descent which is mention in assignment

```python
w = 0
learning_rate = 0.01
gradient = 2 * (w - 5)
w = w - learning_rate * gradient
print("Updated value of w:", w)
```

### 4. Effect of Different Learning Rates on Gradient Descent

```python
import matplotlib.pyplot as plt
learning_rates = [0.1, 0.01, 0.5]
iterations = 15
for lr in learning_rates:    
    x = 0
    x_values = []    
    for i in range(iterations):
        gradient = 2 * (x - 3)      
        x = x - lr * gradient       
        x_values.append(x)    
    plt.plot(
        range(iterations),
        x_values,
        label=f"Learning Rate = {lr}"
    )
plt.xlabel("Iterations")
plt.ylabel("Value of x")
plt.title("Gradient Descent with Different Learning Rates")
plt.legend()
plt.grid(True)
plt.show()
```

### 5. Plot f(x) = x⁴ - 3x³ + 2 and Identify Minima

```python
import numpy as np
import matplotlib.pyplot as plt
def plot_function():
    x = np.linspace(-2, 4, 400)
    y = x**4 - 3*x**3 + 2
    plt.figure(figsize=(8,5))
    plt.plot(x, y, label="f(x)=x⁴-3x³+2")
    local_min_x = 2.25
    local_min_y = local_min_x**4 - 3*local_min_x**3 + 2
    global_min_x = 0
    global_min_y = 2
    plt.scatter(
        local_min_x,
        local_min_y,
        color="red",
        label="Local Minimum"
    )
    plt.scatter(
        global_min_x,
        global_min_y,
        color="green",
        label="Global Minimum"
    )
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title("Function f(x)=x⁴-3x³+2")
    plt.grid(True)
    plt.legend()
    plt.show()
plot_function()
```
