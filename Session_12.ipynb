# 3. Functional API Model

## Part 3 — Code & Experiment

````python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model

inputs = Input(shape=(10,))
x = Dense(20, activation='relu')(inputs)
outputs = Dense(1)(x)

model = Model(inputs=inputs, outputs=outputs)

model.summary()

print("Total Parameters:", model.count_params())
````

# 5. Keras Model with Two Inputs

## Part 5 — Code & Experiment

````python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, Concatenate
from tensorflow.keras.models import Model

# Input 1: User Age
age_input = Input(shape=(1,), name="Age")

# Input 2: Number of Songs Listened
songs_input = Input(shape=(1,), name="Songs_Listened")

# Combine inputs
merged = Concatenate()([age_input, songs_input])

# Hidden layers
x = Dense(8, activation='relu')(merged)
x = Dense(4, activation='relu')(x)

# Output
output = Dense(1, activation='sigmoid', name="Premium_Probability")(x)

# Create model
model = Model(
    inputs=[age_input, songs_input],
    outputs=output
)

model.summary()
````
