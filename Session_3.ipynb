# Session 3 – Basic Neuron Concepts

This notebook demonstrates how a simple artificial neuron works, how weights can be initialized, why bias matters, and how neurons handle multiple input values.

## 1. Build a Simple Neuron Function

```python
def calculate_neuron(feature_value, connection_weight, neuron_bias):
    final_value = (feature_value * connection_weight) + neuron_bias
    return final_value


sample_output = calculate_neuron(
    feature_value=12,
    connection_weight=0.65,
    neuron_bias=1.5
)

print("Neuron output:", sample_output)
```

## 2. Apply the Neuron Function to Social Media Engagement

```python
def calculate_neuron(feature_value, connection_weight, neuron_bias):
    return (feature_value * connection_weight) + neuron_bias


engagement_scores = [18, 27, 42, 33, 55]

connection_weight = 0.7
neuron_bias = 1.25

print("Neuron results based on engagement values:\n")

for score in engagement_scores:
    neuron_result = calculate_neuron(
        score,
        connection_weight,
        neuron_bias
    )

    print(
        f"Engagement: {score} "
        f"=> Neuron Output: {neuron_result:.2f}"
    )
```

## 3. Compare Different Weight Initialization Methods

```python
import numpy as np

number_of_connections = 8

# Strategy 1: All weights start from zero
weights_zero = np.zeros(number_of_connections)

# Strategy 2: Random values within a selected range
weights_random = np.random.uniform(
    low=-0.8,
    high=0.8,
    size=number_of_connections
)

# Strategy 3: Xavier-style normal initialization
xavier_std = 1 / np.sqrt(number_of_connections)
weights_xavier = np.random.normal(
    loc=0,
    scale=xavier_std,
    size=number_of_connections
)

print("Zero initialized weights:")
print(weights_zero)

print("\nRandom initialized weights:")
print(weights_random)

print("\nXavier initialized weights:")
print(weights_xavier)
```

## 4. Understand the Importance of Bias

```python
def calculate_neuron(feature_value, connection_weight, neuron_bias):
    return (feature_value * connection_weight) + neuron_bias


feature_value = 0
connection_weight = 0.75

result_without_bias = calculate_neuron(
    feature_value,
    connection_weight,
    0
)

result_with_bias = calculate_neuron(
    feature_value,
    connection_weight,
    3.5
)

print("When bias is 0   :", result_without_bias)
print("When bias is 3.5 :", result_with_bias)

print("\nThe bias allows a neuron to produce a shifted output even when the input is zero.")
```

## 5. Create a Neuron That Accepts Multiple Inputs

```python
def multi_input_neuron(feature_list, weight_list, neuron_bias):
    total = 0

    for feature, weight in zip(feature_list, weight_list):
        total += feature * weight

    return total + neuron_bias


input_features = [3, 5, 2, 4]

neuron_weights = [
    0.45,
    0.25,
    0.15,
    0.35
]

neuron_bias = 0.8

final_output = multi_input_neuron(
    input_features,
    neuron_weights,
    neuron_bias
)

print("Input features :", input_features)
print("Neuron weights:", neuron_weights)
print("Bias value     :", neuron_bias)
print("Final output   :", final_output)
```
