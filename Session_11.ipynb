### 1. Train a Fashion MNIST Classifier using SGD — Practice

## Part 1 — Implementation

```python
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.optimizers import SGD
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
x_train = x_train / 255.0
x_test = x_test / 255.0
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])
model.compile(
    optimizer=SGD(),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.18
)
```

### 2. Replace SGD with RMSProp

## Part 2 — Implementation

```python
from tensorflow.keras.optimizers import RMSprop
model.compile(
    optimizer=RMSprop(),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.18
)
```

### 3. Adam Optimizer with Learning Rate = 0.0005

## Part 3 — Implementation

```python
from tensorflow.keras.optimizers import Adam
model.compile(
    optimizer=Adam(step_size=0.0005),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.18
)
```

### 4. Learning Rate Scheduler with Momentum Optimizer

## Part 4 — Implementation

```python
from tensorflow.keras.optimizers import SGD
from tensorflow.keras.callbacks import LearningRateScheduler
initial_lr = 0.01
def scheduler(epoch, lr):
    if epoch > 0 and epoch % 2 == 0:
        return lr * 0.47
    return lr
callback = LearningRateScheduler(scheduler)
optimizer = SGD(
    step_size=initial_lr,
    momentum=0.87
)
model.compile(
    optimizer=optimizer,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
history = model.fit(
    x_train,
    y_train,
    epochs=6,
    validation_split=0.18,
    callbacks=[callback]
)
```

# 5. Why Adam is Often Preferred

Adam is one of the most popular optimizers because it combines the advantages of **Momentum** and **RMSProp**. Instead of using the same learning rate for every parameter, Adam automatically adjusts the learning rate for each weight based on past gradients. This usually leads to **faster convergence**, **less manual tuning**, and **better performance** on many deep learning problems. Because of these properties, Adam is widely used for image classification, natural language processing (NLP), and recommendation systems.

## Real-world Example

Consider **Spotify's music recommendation system**. The recommendation model is trained on millions of user interactions. If the optimizer converges slowly (such as plain SGD), training can take much longer and may produce less accurate recommendations. Using Adam allows the model to learn user preferences more efficiently, resulting in more relevant song recommendations and a better listening experience for users.

## Overall Comparison

| Optimizer | Learning Rate | Validation Accuracy (Typical) | Training Speed |
|-----------|---------------|-------------------------------|----------------|
| SGD | Default | ~84% | Slow |
| RMSProp | Default | ~88% | Medium |
| Adam | 0.0005 | ~89–90% | Fast |
| SGD + Momentum + LR Scheduler | Starts at 0.01 | ~88–89% | Stable |

## Conclusion

For this simple Fashion MNIST classifier, **Adam** with a learning rate of **0.0005** typically provides the best balance of training speed and accuracy. **RMSProp** generally outperforms plain **SGD**, while **SGD combined with Momentum and a learning-rate scheduler** offers a strong alternative with stable convergence.
