## 1. Movie Recommendation Neural Network (Two Inputs) — Practice

## Part 1 — Experiment

```python
from tensorflow.keras.layers import Input, Dense, Embedding, Flatten, Concatenate
from tensorflow.keras.models import Model

# Input 1: User age
age_input = Input(shape=(1,), name="Age")

# Input 2: Favorite genre (integer encoded)
genre_input = Input(shape=(1,), name="Genre")

# Convert genre integer into embedding
genre_embedding = Embedding(input_dim=10, output_dim=4)(genre_input)
genre_embedding = Flatten()(genre_embedding)

# Combine inputs
combined = Concatenate()([age_input, genre_embedding])

# Hidden layers
x = Dense(18, activation='relu')(combined)
x = Dense(10, activation='relu')(x)

# Output: Predicted rating (1-5)
rating_output = Dense(1, activation='linear', name="Movie_Rating")(x)

# Create model
model = Model(inputs=[age_input, genre_input], outputs=rating_output)

# Compile
model.compile(optimizer='adam', loss='mse', metrics=['mae'])

# Model summary
model.summary()
```

## 2. Multi-Output Neural Network for Zomato Delivery Prediction

## Part 2 — Experiment

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model

# Inputs
amount = Input(shape=(1,), name="Order_Amount")
distance = Input(shape=(1,), name="Distance")

# Combine features
x = Dense(18, activation='relu')(amount)
y = Dense(18, activation='relu')(distance)

merged = Dense(28, activation='relu')(
    Dense(18, activation='relu')(
        tf.keras.layers.Concatenate()([x, y])
    )
)

# Output 1: Delivery Time
delivery_time = Dense(1, name="Delivery_Time")(merged)

# Output 2: Delivery Rating
delivery_rating = Dense(1, name="Delivery_Rating")(merged)

# Model
model = Model(
    inputs=[amount, distance],
    outputs=[delivery_time, delivery_rating]
)

# Compile
model.compile(
    optimizer='adam',
    loss={
        "Delivery_Time": "mse",
        "Delivery_Rating": "mse"
    },
    metrics={
        "Delivery_Time": "mae",
        "Delivery_Rating": "mae"
    }
)

model.summary()
```

### Functional API Version

## Part 4 — Experiment

```python
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model

# Input layer
inputs = Input(shape=(30,), name="Music_Features")

# Hidden layers
x = Dense(48, activation='relu')(inputs)
x = Dense(28, activation='relu')(x)

# Output layer
outputs = Dense(12, activation='softmax', name="Playlist")(x)

# Functional model
model = Model(inputs=inputs, outputs=outputs)

# Compile
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

## 4. Skip Connection for IPL Cricket Score Predictor

## Part 5 — Experiment

```python
from tensorflow.keras.layers import Input, Dense, Add
from tensorflow.keras.models import Model

# Input features
inputs = Input(shape=(10,), name="Match_Features")

# Hidden path
hidden = Dense(18, activation='relu')(inputs)
hidden = Dense(1, activation='linear')(hidden)

# Direct shortcut path
shortcut = Dense(1, activation='linear')(inputs)

# Skip connection
output = Add(name="Predicted_Score")([hidden, shortcut])

# Model
model = Model(inputs=inputs, outputs=output)

# Compile
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

model.summary()
```
