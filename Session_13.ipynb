## Part 1 — Experiment

```python
import numpy as np
from tensorflow.keras.classifiers import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
from scikeras.wrappers import KerasRegressor

# For reproducible results
np.random.seed(32)

# -----------------------------
# Simulate dataset
# -----------------------------
# Features:
# listening_time (0-100)
# liked_genre_score (0-1)
listening_time = np.random.rand(100) * 100
liked_genre_score = np.random.rand(100)

X = np.column_stack((listening_time, liked_genre_score))

# Binary labels (0 or 1)
y = np.random.randint(0, 2, size=100)

# -----------------------------
# Build the classifier
# -----------------------------
classifier = Sequential([
    Dense(20, activation='relu', input_shape=(2,)),
    Dense(6, activation='relu'),
    Dense(1, activation='sigmoid')
])

# -----------------------------
# Compile the classifier
# -----------------------------
classifier.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# -----------------------------
# Early Stopping Callback
# -----------------------------
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

# -----------------------------
# Train the classifier
# -----------------------------
history = classifier.fit(
    X,
    y,
    epochs=15,
    validation_split=0.17,
    callbacks=[early_stopping],
    verbose=1
)

# -----------------------------
# Evaluate and print accuracy
# -----------------------------
loss, accuracy = classifier.evaluate(X, y, verbose=0)

print(f"\nTraining Accuracy: {accuracy:.4f}")
```
