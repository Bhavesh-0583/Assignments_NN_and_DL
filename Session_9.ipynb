### 2. Observe Weight Updates Before and After Training

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

network = keras.Sequential([
    layers.Dense(5, activation="relu", input_shape=(3,)),
    layers.Dense(1, activation="sigmoid")
])

features = np.array([
    [2, 1, 4],
    [5, 3, 7],
    [1, 4, 2]
], dtype=float)

targets = np.array([
    [0],
    [1],
    [0]
], dtype=float)

print("Weights before training:")
for layer_number, layer in enumerate(network.layers, start=1):
    print(f"Layer {layer_number}:", layer.get_weights())

network.compile(
    optimizer="sgd",
    loss="binary_crossentropy"
)

network.fit(features, targets, epochs=1, verbose=0)

print("\nWeights after training:")
for layer_number, layer in enumerate(network.layers, start=1):
    print(f"Layer {layer_number}:", layer.get_weights())
```

## 3. Manual Gradient Calculation Using the Chain Rule

### Given Values

- Input: \(x = 3\)
- Weight: \(w = 0.4\)
- Bias: \(b = 0.8\)
- Target: \(y = 1\)
- Activation Function: Sigmoid

---

## Step 1: Calculate the Weighted Sum

The value before activation is:

$$
z = wx + b
$$

Substitute the values:

$$
z = (0.4 \times 3) + 0.8
$$

$$
z = 1.2 + 0.8
$$

$$
\boxed{z = 2}
$$

---

## Step 2: Apply the Sigmoid Activation

The sigmoid function is:

$$
a = \frac{1}{1 + e^{-z}}
$$

For \(z=2\):

$$
a = \frac{1}{1 + e^{-2}}
$$

Therefore:

$$
\boxed{a \approx 0.8808}
$$

---

## Step 3: Binary Cross-Entropy Loss

The loss is:

$$
L = -[y\log(a) + (1-y)\log(1-a)]
$$

Because \(y=1\):

$$
L = -\log(a)
$$

---

## Step 4: Gradient with Respect to the Weight

For sigmoid combined with binary cross-entropy:

$$
\frac{\partial L}{\partial w} = (a-y)x
$$

Substitute the values:

$$
\frac{\partial L}{\partial w}
= (0.8808 - 1)\times3
$$

$$
= (-0.1192)\times3
$$

$$
\boxed{\frac{\partial L}{\partial w} \approx -0.3576}
$$

---

## Final Result

The gradient is approximately:

$$
\boxed{\frac{\partial L}{\partial w} = -0.3576}
$$

### Interpretation

The negative sign means that, for this example, moving the weight upward would move the prediction toward the target and reduce the loss.

### 4. Simulate Vanishing Gradient Using Repeated Sigmoid

```python
import math

def sigmoid(value):
    return 1 / (1 + math.exp(-value))

signal = 0.7
depth = 12

print("Starting value:", signal)

for layer_index in range(1, depth + 1):
    signal = sigmoid(signal)
    print(f"Layer {layer_index:02d}: {signal:.6f}")

print("\nFinal output:", round(signal, 6))
```

### 5. Demonstrating Exploding Gradients with Larger Weights

```python
import numpy as np

weight_matrix = np.array([
    [12, 12],
    [12, 12]
])

input_vector = np.array([4, 4])

activation = np.dot(input_vector, weight_matrix)

print("Activation values:")
print(activation)

gradient_value = activation.astype(float)

for _ in range(5):
    gradient_value *= weight_matrix.mean()

print("\nGradient after repeated backpropagation:")
print(gradient_value)
```
